```python
import requests
from bs4 import BeautifulSoup
url1="https://subslikescript.com/movies"
response=requests.get(url1) #step 1

print(response.status_code)
```

```python
import requests
from bs4 import BeautifulSoup
url1="https://www.politifact.com/"
response=requests.get(url1) #step 1

print(response.status_code)
```

```python
response=response.text #step 2
soup=BeautifulSoup(response,"html.parser")
#print(soup)
#print(soup.prettify())
```

```python
source=[]
statement=[]
link1=[]
date4=[]

a=soup.find("div",class_="embla")
b=a.find_all("div",class_="embla__slide caro-listing__item")
for i in b:
    post=i.find("div",class_="m-statement__meta").find("a").text.strip()
    source.append(post)
source
```

```python
for i in b:
    st=i.find("div",class_="m-statement__quote-wrap").find("a").text.strip()
    statement.append(st)
statement
```

```python
for i in b:
    li=i.find("div",class_="m-statement__quote").find("a")["href"]
    link1.append("https://www.politifact.com/"+li)
link1
```

```python
for i in b:
    d=i.find("div",class_="m-statement__desc").text.strip()
    l_d=d.split()
    l1=(" ".join(l_d[2:5]))
    date4.append(l1)
date4
```

```python
import pandas as pd 

df=pd.DataFrame({"source":source,"statement":statement,"link":link1,"date":date4})
df
```

```python
import requests
from bs4 import BeautifulSoup
website="Certified used Mercedes-Benz for sale.html"
file=open(website,encoding="utf-8") #step1
response=file.read() #step2 
soup=BeautifulSoup(response,"html.parser")
print(soup)
```

```python
car_name=[]
rating=[]
review_count=[]
price=[]
dealer_name=[]
mileage1=[]

a=soup.find_all("div",class_="vehicle-details")
for i in a:
    post=i.find("h2",class_="title").text.strip()
    car_name.append(post)
car_name
```

```python
for i in a:
    m=i.find("div",class_="mileage").text.strip()
    mileage1.append(m)
mileage1
```

```python
for i in a:
    d=i.find("div",class_="dealer-name").text.strip()
    dealer_name.append(d)
dealer_name
```

```python
for i in a:
    p=i.find("span",class_="primary-price").text.strip()
    price.append(p)
price
```

```python
for i in a:
    review_count.append(i.find("span",class_="sds-rating__link sds-button-link").text.strip()[1:-9])
review_count
```

```python
import numpy as np
for i in a:
    try:
        rating.append(i.find("span",class_="sds-rating__count").text.split())
    except:
        rating.append(np.nan)
rating
```

```python
import pandas as pd
df=pd.DataFrame({"Car Name":car_name,"rating":rating,"review_count":review_count,"price":price,})
```

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
website="30 Most Popular Movies Right Now_ What to Watch In Theaters and Streaming __ Rotten Tomatoes – Movie and TV News.html"
file=open(website,encoding="utf-8") #step1
response=file.read() #step2 
soup=BeautifulSoup(response,"html.parser")

movie=[]
rating=[]
director=[]
a=soup.find_all("div",class_="col-sm-18 col-full-xs countdown-item-content")

for i in a:
    movie.append(i.find("h2").find("a").text.strip())
    rating.append(i.find("span",class_="tMeterScore").text.strip())
    director.append(i.find("div",class_="info director").find("a").text.strip())

df=pd.DataFrame({"movie":movie,"rating":rating,"director":director})
df
```

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
website="indeed jobs (1).html"
file=open(website,encoding="utf-8") #step1
response=file.read() #step2 
soup=BeautifulSoup(response,"html.parser")

name_job=[]
name_company=[]
job_location=[]
job_details=[]

a=soup.find_all("div",class_="job_seen_beacon")

for i in a:
    name_job.append(i.find("h2").find("a").text.strip())
    name_company.append(i.find("span",class_="css-1h7lukg eu4oa1w0").text.strip())
    job_location.append(i.find("div",class_="css-1restlb eu4oa1w0").text.strip())
    job_details.append(i.find("div",class_="underShelfFooter").text.strip())

df=pd.DataFrame({"name-of-the-job":name_job,"name_company":name_company,"job_location":job_location,"job_details":job_details})
df
```

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
website="SENSEX Top Gainers, BSE Daily Top Gainers - Stock Market (1).html"
file=open(website,encoding="utf-8") #step1
response=file.read() #step2 
soup=BeautifulSoup(response,"html.parser")

company=[]
current_price=[]
prev_closedate=[]
Change=[]
Change_pre=[]
Low_high=[]

a=soup.find("table",class_="common-table")
tr=a.find_all("tr")

for i in tr[1:]:
    company.append(i.find("td",class_="v-align").text.strip())
    current_price.append(i.find("td",class_="txt-right v-align").text.strip())
    prev_closedate.append(i.find("td",class_="txt-right no-wrap").text.strip())
    Low_high.append(i.find("span",class_="fl f-9").text.strip())
    C=i.find_all("td",class_="txt-ctr")
    for i in C[::2]:
        Change.append(i.text.strip())
    for i in C[1::2]:
        Change_pre.append(i.text.strip())
Low_high
df=pd.DataFrame({"COMPANY":company,"CURRENT PRICE":current_price,"PREVIOUS CLOSE DATE":prev_closedate,"CHANGE%":Change,"CHANGE%":Change_pre,"Low_High":Low_high})
df
```
